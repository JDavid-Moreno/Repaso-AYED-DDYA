# Merge Sort


Para este algoritmo necesitamos 2 funciones:

la primera se encarga de recibir la lista y dividirla en izquierda y derecha, y se realiza el mismo procesohasta que cada elemento quede en su propia sublista, una vez separados se retornan las listas al siguiente metodo. 

```python
def merge(array):
    if len(array) == 1:
        return array
    half = len(array) // 2
    left = array[:half]
    right = array[half:]

    sorted_left = merge(left)
    sorted_right = merge(right)

    return merge_sort(sorted_left, sorted_right)
```

el segundo se encarga de crear una nueva lista donde se ordenaran los elementos, de igual forma esta recibe ambas listas llamadas previamente izquierda y derecha, las cuales vamos a verificar que no sean nulas (osea su longitud sea mayor que 0), una vez verificado procedemos a comparar cual es menor, para irla agregando a la nueva lista e ir borrando el elemento agregado de la lista donde se encontraba, en caso de que alguna de las dos listas se vuelva nula al borrar el elemento agregado a la nueva lista, cambiamos de ciclo y miramos cual lista no quedo nula para agregar ese ultimo valor faltante a la lista.

Finalmente, cuando ya todas las listas sean nulas, retornamos la nueva lista la cual tendra todos los elementos originales ya ordenados.

```python
def merge_sort(left, right):
    array_sort = []
    while len(left) > 0 and len(right):
        if left[0] > right[0]:
            array_sort.append(right[0])
            right.pop(0)
        else:
            array_sort.append(left[0])
            left.pop(0)

    while len(left) > 0:
        array_sort.append(left[0])
        left.pop(0)
    while len(right) > 0:
        array_sort.append(right[0])
        right.pop(0)

    return array_sort
```
