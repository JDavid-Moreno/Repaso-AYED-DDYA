# Quick Sort

Este algoritmo al igual que el merge utiliza dos funciones:

La primera consiste en manejar la recursion, asi como, una vez se haga la primera iteracion esta se encarge de volver a realizar dicha iteracion, sin embargo, ya sin el pivote, es decir, que haga quick sort a los sub arreglos de los menores y mayores respecto al pivote, que ya quedaron cada uno en su sitio.

```python
def quick_sort(array, low, high):
    if low < high:
        part = partition(array, low, high)

        quick_sort(array, low, part - 1)
        quick_sort(array, part + 1, high)
```

La segunda se encarga de realziar las particiones para que la funcion `quick_sort` reliace nuevamente la recursion utilizando esas particiones, para esto la funcion asigna el pivote el cual para este caso sera el ultimo elemento del arreglo, asi mismo, creamos la variable `i` que sera nuestro apuntador, el cual empezara en la posicion -1, esto para que nunca vaya delante de la iteracion y que en caso de que el apuntador no se mueva, al momento de mover el pivote quede en la primera posición, y `j` sera quien vaya avanzando en la lista. 

Una vez `j` comience a recorrer toda la lista, verificara que si el elemento en el que esta parado es menor o igual al pivote este mueve el apuntador y hace que la posición del apuntador y la del elemento al que esta parado intercambien.
en el caso es que el elemento que va a ser intercambiado de posicion no tuvo ningun elemento mayor al pivote previo, este se quedara en su misma posición, por ejemplo:


`[1,3,2]` el pivote es 2 y el apuntador empieza en -1, al iterar se compara con el primer elemento osea 1, en ese caso al ser menor el apuntador se mueve a la posición 0 y se hace el intercambio, sin embargo al estar tanto `j` como el apuntador en 0, este no se mueve de posición.

por otro lado si el elemento es mayor, el apuntador se queda justo donde esta, por lo que si hubo un elemento mas grande que el pivote y despues uno mas pequeño entonces intercambian, por ejemplo: 

`[4,1,2,3]` aqui el pivote es 3 y el apuntador empieza en -1, comenzamos a iterar y vemos que 3 > 4 por lo que el puntero no se mueve, sigue avanzando `j` a 1 y 3 > 1 por lo que el apuntador se mueve a la posición 0, por lo que se hace el intercambio, sin embargo, al ir `j` en 2, al hacer el intercambio queda:
`[1,4,2,3]`, y asi se va haciendo mientras hasta finalizar la iteración, esto se hace para que los elementos mas grandes que el pivote que quedaron antes que los elementos mas pequeños que este se vayan se vayan moviendo para que queden los mas pequeños a la izquierda y los mas grandes a la derecha.

Finalmente, al acabar la iteración, se intercambia el elemento que va justo despues del apuntador con el pivote, ya que lso elementos que van desde el apuntador hacia atras son mas pequeños que el pivote y los elementos que van despues del apuntador son mas grandes que el pivote, es por eso que el apuntador empieza en -1, ya que si no se encuentran elementos mas pequeños, el pivote se va a la primera posición. 

```python
def partition(array, low, high):
    pivot = array[high]
    i = low - 1
    for j in range(low, high):
        if array[j] <= pivot:
            i = i + 1
            array[i], array[j] = array[j], array[i]
    array[i + 1], array[high] = array[high], array[i + 1]
    return i + 1
```

```python
def main():
    array = [4,1,3,8,5,7,2,6]
    quick_sort(array, 0, len(array) - 1)
    print(array)
main()
```
