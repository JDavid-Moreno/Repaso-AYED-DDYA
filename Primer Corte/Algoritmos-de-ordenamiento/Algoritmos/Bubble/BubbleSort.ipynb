# Bubble Sort

Implementacion de codigo para el algoritmo de ordenamiento Bubble Sort.

```python
def bubblesort(array):
    n = len(array)
    for i in range(n):
        for j in range(0, n-i-1):
            if array[j] > array[j+1]:
                array[j], array[j+1] = array[j+1], array[j]
    return array
```

```python
def main():
    array = [4,1,3,8,5,7,2,6]
    print(bubblesort(array))
main()
```

Primero se genera la lista desordenada, sin embargo si se quiere verificar antes que la lista ya esta ordenada, una manera sencilla seria usar una verificacion usando `sorted()`, pero esto genera un problema porque esta funcion ordena la lista lo que genera una complejidad de `O(n log(n)) innecesaria, por lo que si se quiere verificar la funcion quedaria.

```python
def bubblesort(array):
    if array == sorted(array):
        return array
    n = len(array)
    for i in range(n):
        for j in range(0, n-i-1):
            if array[j] > array[j+1]:
                array[j], array[j+1] = array[j+1], array[j]
    return array
```

Por otro lado, su funcionamiento consiste en definir dos ciclos `i` y `j`, donde j va desde 0 osea el principio de la lista hasta (n-i-1) que seria hasta el ultimo elemento ordenado, esto para que no revise los ultimos elementos los cuales ya estan ordenados, asi mismo, j se usa para verificar dos elementos, un elemento n y su elemento siguiente (n + 1), los cuales se comparan, en donde si el primer elemento es mayor al segundo, estos cambien de lugar. Por otro lado, i va por todo el arreglo y se encarga que j vaya avanzando y verificando cada pareja de elementos.
