# Selection Sort

Se utilizan dos ciclos for con las variables `i` y `j`. El elemento en la posición `i` se define inicialmente como el mínimo en cada iteración, hasta que `j` encuentre un valor más pequeño; en ese caso, el índice del mínimo cambia al del elemento `j` encontrado. Este proceso se repite en caso de hallar otro elemento `j` aún menor. Una vez que `j` recorre toda la lista y se determina el valor mínimo real de esa iteración, los elementos en la posición `i` y en la posición del mínimo se intercambian para que este último quede en su respectiva posición.

Una vez realizado esto, se vuelve a iterar incrementando `i` e ignorando los elementos ya ordenados, ya que, al estar en su posición correcta, no es necesario validarlos y sería redundante.

```python
def selection(array):
    n = len(array)
    for i in range(n):
        minium = i
        for j in range(i+1, n):
            if array[j] < array[minium]:
                minium = j
        if minium != i:
            array[i], array[minium] = array[minium], array[i]
    return array
```

```python
def main():
    array = [4,1,3,8,5,7,2,6]
    print(selection(array))
main()
```
