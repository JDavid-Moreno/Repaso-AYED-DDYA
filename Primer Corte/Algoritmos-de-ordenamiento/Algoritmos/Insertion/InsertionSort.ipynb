# Insertion Sort

Para este algoritmo a diferencia de los anteriores se utiliza un for y un while, por lo su mejor caso es de complejidad O(n) a diferencia de los anetriores que tanto su mejor como peor caso es O(n²), para este `i` empieza desde 1 en vez desde 0 osea el principio de la lista, esto ya que asumimos que el primer elemento esta ordenado ya que no tiene elementos previos para comparar, por otro lado definimos una variable `key`, esta variable la podemos imaginar con el ejemplo de las cartas, tengo en una mano cartas y las quiero ordenar una por una, con la otra mano agarro la carta que quiero ordenar y la pongo donde deberia ir, eso es key, la carta que vamos a mover de posición. 

Posteriormente, definimos la variable `j` la cual es la encarga de mover los elementos hacia atras en caso de que el elemento `key` sea menor, por esa razon se tiene en el while la condicion `j > 0` ya que en caso de que `j` sea 0 significa que llegamos al inicio de la lista por lo que podemos inserta el elemento en esa posición ya que es el menor valor que ahi en la lista.

```python
def insertion(array):
    n = len(array)
    for i in range(1, n):
        key = array[i]
        j = i
        while j > 0 and array[j - 1] > key:
            array[j] = array[j - 1]
            j -= 1
        array[j] = key
    return array
```

```python
def main():
    array = [4,1,3,8,5,7,2,6]
    print(insertion(array))
main()
```
